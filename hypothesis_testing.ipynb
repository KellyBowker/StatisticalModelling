# Hypothesis testing


> A Dutch nurse took a bad feeling to the chief pediatrician. A six-month-old baby had died overnight at the brick children’s hospital in The Hague, in September 2001. The death was ruled natural. The nurse worried about who had been on duty: Lucia de Berk, again.

The quote above is from an essay on the hypothesis test [analysis](https://statistical.systems/essays/significant_but_wrong/) of the case of [Lucia de Berk](https://en.wikipedia.org/wiki/Lucia_de_Berk_case). In this case, a pediatric nurse was convicted of murdering several babies. If you have the inclination and the time, I recommend reading or listening to this [collection of essays](https://statistical.systems/essays/tests/) at some future time (though it may be a bit abstract). I would also recommend this [TED talk](https://www.youtube.com/watch?v=kLmzxmRcUTo) for building greater statistical intuition and a discussion on Lucia's case. For now let's go on with the story. 

Lucia had been on call several times when there had been an incident (a need for medical intervention or death) with babies in the maternity ward of the children's hospital where she worked. It later turned out she had been present during incidents at other wards in the Red Cross hospital. Evidence against her rested on this table:
| Ward | Number of shifts | Lucia's shifts | Number of incidents | Number of incidents on Lucia's shifts |
| --------------------|------|-----|----|---| 
| Children's hospital | 1029 | 142 | 8  | 8 |
| Red Cross, ward 41  | 366  | 1   | 5  | 1 | 
| Red Cross, ward 42  | 339  | 58  | 14 | 5 | 

So, is Lucia de Berk guilty or is her presence during the incidents due to chance?

We could make the following assumptions:
1. Every nurse has the same chance of an incident on any given shift
2. Every shift is independent from one another
3. The wards are independent from one another

In this case, you could work out the probability of one nurse being present in all these incidents be 1 in 342 million (see [this](https://statistical.systems/essays/significant_but_wrong/) for the full math). However...

1. Nurses with different training, get different shifts - rates of incidents are not the same for everyone
2. What happens in one shift can affect the following ones - shifts are not independent
3. Wards have different practices and receive different patients depending on their expertise - they cannot be treated as independent

If you take these things into account (also in this [paper](https://arxiv.org/abs/1009.0802)), it comes out that the chance of Lucia being present in 14 out of 27 total incidents is 1 in 49. 

Lucia was initially convicted to life imprisonment in 2004 but later her conviction was overturned in 2010. 

Answering questions like these is the topic of this week. Not just **"is there a relationship?"**, in the case of Lucia there was!, she was a nurse working at a place where incidents are likely to occur. Hypothesis testing is about going further and asking **"is this relationship due to chance?"**. When answering the second question it is easy to make faulty assumptions on, like in the case of Lucia de Berk.

The first step in hypothesis testing is defining the hypothesis. A **null hypothesis (H$_0$)** is established and it is common practice to establish an **alternative hypothesis (H$_1$)**.
1. A null hypothesis is a statement about there being **no significant** differences or relationship between two samples, i.e. observations are due to chance.
2. An alternative hypothesis (H$_1$) can be the negation of the null hypothesis (H$_0$), but it can also take another form. 

Say we are comparing performance between male and female university students.
An example could be:
1. H$_0$: there is **no** difference between the performance of male and female university students.
2. H$_1$: female students perform better academically in university.

Note in the example above that the alternative hypothesis is not only the negation of the null, it is also a statement about the relationship between these two variables. If we just wanted to negate H$_0$ it could be said "H$_1$: there is a difference between the performance of male and female university students".

The second step is picking a **[test statistic](https://en.wikipedia.org/wiki/Test_statistic)**. A test statistic is computed from the data and is used to compute the p-value: the **conditional probability** of an observation being **at least** this extreme, supposing the null hypothesis is true. 

More intuitively, the p-value is the probability of, assuming H$_0$ is true, observing a result that contradicts H$_0$ just as much, or more, than the case we want to test does. 

A lot to unpack. The test statistic is a method for computing a **p-value**. The p-value **is not** the probability of H$_0$ being true. It is the probability of observing values **as extreme** or **more extreme** than the ones we actually have sampled, **assuming** the null hypothesis is true. Then, we can interpret the p-value
1. If the p-value is above a certain threshold (scientific field dependent), we fail to reject the null hypothesis
2. If the p-value is below that threshold, the null hypothesis is rejected 
3. If the p-value is similar to the threshold, we have to defer judgement about the null hypothesis

Because there is no universal way of establishing a threshold, we cannot say that the null hypothesis is **false** if we reject it. The reverse is also true: it is impossible to **accept** a null hypothesis, we can only fail to reject it. 

It is also useful to distinguish between two possible classes of error:
1. Type I error, in which H$_0$ is rejected despite it being true.
2. Type II error, in which we fail to reject H$_0$, despite H$_0$ being false

We will make several distinctions regarding test statistics in this notebook, but there are so many to explore [here](https://en.wikipedia.org/wiki/Test_statistic#Common_test_statistics) or [here](https://docs.scipy.org/doc/scipy/reference/stats.html#hypothesis-tests-and-related-functions).
1. A one-sample test is appropriate when a sample is compared to a target. Example: is the  waiting time at a hospital sufficiently low. (e.g. H$_0$: $t_{wait} < 5$ hours)
2. A two-sample test is appropriate when comparing two different samples. Example: experimental vs control groups in a medical trial. (e.g. H$_0$: $\mu_{control}=\mu_{experimental}$)
3. A $t$-test is appropriate when comparing means. Example: the average score of students in university.
4. A F-test appropriate when comparing variances. Example: is there more variation of penguin flipper lengths in one island vs another.
5. A $\chi^2$-test of independence is appropriate for comparing if two categorical variables are independent. Example: if the category species of penguins are associated with certain antarctic islands.

Note that the many tests **assume** normal distributions for samples and errors. If you have a lot of samples, the [Central Limit Theorem](https://en.wikipedia.org/wiki/Central_limit_theorem) allows us to approximate the sampling distribution of the mean by a normal distribution. There is no precise $n$ for which this applies, it heavily depends on the underlying samples. There are two other strategies to deal with populations which are non-normal, we will look into this later. 

The above tests cover a lot of use cases, but it might not cover all. It can also be that you need to chain more than one test to make a sound analysis. Note also that some tests make assumptions that might not be true in your case. Always check the documentation to make sure a test statistic is appropriate for you. If you choose a test but are not sure if it works for your case, let us know. For now, let's see some examples.

We will start with $t$-test examples. As a first example, let's test whether the average of flipper lengths is the statistically the same for two different species of penguins. The easiest way to do this is to do a **two-sample t-test**. The test we are about to perform assumes that these distributions have equal sample variances. To apply this correctly we would have to check that, but for now we will just proceed with the $t$-test and pass an option to the function such that this assumption is bypassed.

To be specific:
- H$_0$: the mean of flipper lengths of Adelie and Gentoo penguins is the statistically same
- H$_1$: the mean of flipper lengths of Adelie and Gentoo penguins is not the statistically same

````python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats
import random

df = pd.read_csv("penguins_size.csv")
df = df.dropna(how='any')

adelie = df[df['species']=='Adelie']
gentoo = df[df['species']=='Gentoo']
chinstrap = df[df['species']=='Chinstrap']

a = adelie['flipper_length_mm'].to_numpy()
b = gentoo['flipper_length_mm'].to_numpy()
print(stats.ttest_ind(a,b,equal_var=False))
````

We obtained a p-value on the order of 10$^{-95}$, which is quite close to zero. That means we can reject the null hypothesis H$_0$ and accept the alternative: the mean of flipper lengths of Adelie and Gentoo species is not the same. This was expected from our prior knowledge of this dataset.

We already knew that different species of penguins have different physical measurements. But what about between genders of the same species?

As an excuse to do a one-sample test in addition to a two-sample test, let's assume that the theoretical distribution of flipper lengths of female adelie penguins is the same as that of males. Let's also specify:
- H$_0$: the mean of flipper lengths of male and female Adelie penguins is the statistically same
- H$_1$: the mean of flipper lengths of male and female Adelie penguins is not statistically the same

````python
a = adelie[adelie['sex']=='FEMALE']['flipper_length_mm']
b = adelie[adelie['sex']=='MALE']['flipper_length_mm']

mean_males = np.mean(b)
print(stats.ttest_1samp(a,mean_males)) # comparing Female mean flipper length with mean_males
print(stats.ttest_ind(a,b,equal_var=False)) # comparing mean Female and Male flipper length
````

Note that the p-values are quite different! The reason for this difference is that in the one-sample case, we are treating the male mean as a theoretical authoritative number. 

However, the average of the males is its own measurement with associated uncertainties. You can only use another sample's average as a known constant if its uncertainty is negligible compared the uncertainty of your other measurement.

As a final attempt demonstration of the $t$-test, let's use two different datasets. Our hypothesis are:
- H$_0$: the average quality of red and white wines are similar
- H$_1$: the average quality of red and white wines are different

````python
white = pd.read_csv('winequality-white.csv').dropna(how='any')
red = pd.read_csv('winequality-red.csv').dropna(how='any')

white_qual = white['quality'].to_numpy(dtype=float)
red_qual = red['quality'].to_numpy(dtype=float)

print(np.mean(white_qual))
print(np.mean(red_qual))
print(stats.ttest_ind(white_qual,red_qual,equal_var=False))
````

As you can see from the code cell above, the average quality of white and red wines are approximately 5.8779 and 5.6360, respectively. Are these values so close due to chance? Or do humans rate white and red wines on average the same value?

The p-value is very small: $8.17\times 10^{-24}$, which allows us to **reject** the null hypothesis. White wines are rated higher than red wines on average, even if the difference in average ratings is very small. 

How can we be so sure that the wine quality average is different for red and white wines? The numbers are quite similar! The short answer is the size of the datasets. The more samples a dataset has, the more robust the findings and p-values can be. A longer demonstration of uncertainty will be made towards the end of this notebook demo.

The following are two options for comparing variances. The ```levene``` method is robust against non-normal distributions while the ```bartlett``` is not. Our hypothesis are:
- H$_0$: the variances of the quality of red and white wines are the same
- H$_1$: the variances of the quality of red and white wines are not the same

````python
print(stats.levene(white_qual,red_qual,center='mean'))
print(stats.bartlett(white_qual,red_qual))
````

The p-values of the Levene and Bartlett methods are 0.43 and $8.29\times10^{-6}$, respectively. 

Under the Levene test, we fail to reject the null hypothesis.

We dismiss the result of the Bartlett test because it is not robust against a non-normal distribution (wine rating probably has a non-normal distribution).

To compare these tests with actual normal distributions, let's make a small experiment, as a sanity test, using synthetic data:

````python
a = np.random.normal(0, 1, 100)
b = np.random.normal(0, 1, 200)

print(stats.levene(a,b,center='mean'))
print(stats.bartlett(a,b))
````

The code above creates two normal distributions with equal variances but unequal sizes. We use the same tests as before and obtain similar p-values between them, this is just a quick sanity check to verify that Levene and Bartlett work as statistical tests. 

 Let's now test whether male and female penguins have the same variance across some physical measurements.
- H$_0$: the variances of physical characteristics of male and female Gentoo penguins is the same
- H$_1$: the variances of physical characteristics of male and female Gentoo penguins is not the same

````python
gent_m = gentoo[gentoo['sex']=='MALE']
gent_f = gentoo[gentoo['sex']=='FEMALE']

char = ['flipper_length_mm','culmen_length_mm','culmen_depth_mm','body_mass_g']

for c in char:
    a = gent_m[c]
    b = gent_f[c]
    print(c + ": " + str(stats.levene(a,b).pvalue))
````

In the cell of code above, the variances of each available body measurement for Gentoo penguins is compared through a Levene test. The p-values which are obtained range from 0.01 to 0.44 with three of them being above 0.05 (a common threshold for p-values). For culmen length and body mass, we fail to reject the null hypothesis. For culmen depth and flipper length it is a tougher call, but it would not be controversial to reject the null hypothesis for flipper length while deferring judgement on the culmen depth (meaning, rejecting the null hypothesis is threshold dependent and a threshold of p=0.05 is typical for many fields).  Is it the same for Adelie and Chinstrap penguins?

````python
species = [adelie,chinstrap]
char = ['flipper_length_mm','culmen_length_mm','culmen_depth_mm','body_mass_g']

for spe in species:
    print("Species: "+list(spe['species'])[0])
    m = spe[spe['sex']=='MALE']
    f = spe[spe['sex']=='FEMALE']
    
    for c in char:
        a = m[c]
        b = f[c]
        print(c + ": " + str(stats.levene(a,b).pvalue))
    print("\n")
````

For each species, we fail to reject and defer judgement about the null hypothesis for different body characteristics. This is indirectly related to [sexual dimorphism](https://en.wikipedia.org/wiki/Sexual_dimorphism) in animals. In fact, this dataset comes from a [paper](https://doi.org/10.1371/journal.pone.0090081) analyzing sexual dimorphism and its relationship to foraging habits in penguins. 

We have now established tests about means and variances. Linear regression can be used to establish if there is a linear relationship between two continuous variables. The next test is a conceptual evolution from testing if there is a linear relationship between variables. This is the [Chatterjee Xi](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.chatterjeexi.html) test. For this test, We will come back to one of our previously seen examples. In hypothesis terms:
- H$_0$: the variables body mass and flipper lengths of penguins are independent from one another
- H$_1$: the variables body mass and flipper lengths of penguins are dependent from one another

````python
a = gentoo['body_mass_g']
b = gentoo['flipper_length_mm']

print(stats.chatterjeexi(a,b,y_continuous=True))
````

The code above shows a very general and robust test for the null hypothesis that two variables are **not** related. With such a low p-value, we can confidently reject the null hypothesis and conclude: body masses and flipper lengths in penguins are related.

The next level of test is whether two categorical variables in a table are related or not, this can be done with the [$\chi^2$ contigency](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.chi2_contingency.html) test. This test is quite useful in clinical trials, but it expects all counts to be above 5. A further example of this can be found in [here](https://docs.scipy.org/doc/scipy/tutorial/stats/hypothesis_chi2_contingency.html). In this example, we will hypothesize:
- H$_0$: the effect of gender and island of Adelie penguins are equivalent in explaining their population distribution
- H$_1$: the effects of gender and island of Adelie penguins are associated in explaining their population distribution

We can visualize this in a table:

|           | Male | Female |
|-----------|------|--------|
| Biscoe    |  22  |   22   | 
| Dream     |  27  |   28   | 
| Torgersen |  24  |   23   | 

Adelie penguins are quite evenly distributed across islands and genders. So, the p-value is expected to be far from zero i.e. these categories should be equivalent in explaining Adelie penguins' population.

````python
obs = pd.crosstab(adelie['island'], adelie['sex'])
print(stats.chi2_contingency(obs))
````

The p-value in this case is quite high $\approx 0.98$ which makes us fail in rejecting the null hypothesis. The case above is quite dull since Adelie penguins are very well distributed over these islands. Let's check one more example of a survey of hair and eye color in students of the University of Delaware from a survey in the 1970s. The hypothesis are:
- H$_0$: hair and eye colors are independent variables
- H$_1$: hair and eye colors are not independent variables

````python
students = pd.read_csv("HairEyeColor.csv")
print(students.head(12))
print(len(students))
````

The dataset actually contains three categorical variables: hair, eye, gender. The test we are using allows the use of more than one category, but some categories have less than 5 counts. Therefore, we will merge parts of the dataset and remove the dependency on gender. The dataset contains 16 pairs of different hair and eye colors:

````python
freq = students['Freq']
freq_ = freq[0:16].to_numpy()+freq[16:].to_numpy() # black hair,brown eyes Male added to black hair, brown eyes Female; Brown hair, brown eye Male added to brown hair, brown eye female; etc.

obs = np.reshape(freq_,shape=(4,4))
print(np.all(obs>=5)) ## check if number of samples > 5 for all cases
print(stats.chi2_contingency(obs))
````

In the code cell above, we transform the array of frequencies where each row represents an eye color and each column represents a hair color. We also print out a sanity check ```np.all(obs>=5)``` prints True if all entries in the matrix have 5 or more counts. The p-value for this test is very small, which leads us to reject the null hypothesis, hair and eye color are related. This test cannot tell us what that relationship is or looks like.

### An example 
The phycologist Muriel Bristol claimed that she was able to differentiate by taste when tea or milk had been poured first in a cup. Statistician Ronal Fisher decided to put that to the [test](https://en.wikipedia.org/wiki/Lady_tasting_tea). How many times would someone need to succeed on a test to convince you of such a claim? The test consisted of eight cups of tea that were served in random order, four with milk poured first, four with tea poured first. Muriel got all of them right. If she were guessing randomly, the odds of doing so would have been 1 in 70. The p-value of these samples is 1/70$\approx$0.0143.

We will simulate the throw of a dice and test whether the observed statistics are from a fair dice where each face is equally likely to land as any other:
- H$_0$: all faces of the dice are equally likely, i.e. the dice is fair
- H$_1$: the dice is unfair

More than just running his test once, we will run it several times in a loop. Whenever the p-value is smaller than 0.01 (lower than the common threshold 0.05), we will print it:

````python
faces = 6
tries = 100
trials = faces*tries
n_samples = 2000

for j in range(0,trials): #we make this test n_trials times 6*100=600
    a = np.random.randint(faces,size=n_samples) # we generate 2000 random numbers, each random number can have a value between 0 and 5 (i.e. <6)
    
    freq = list()
    for i in range(0,faces):
        freq.append(np.sum(a==i)) # for each face, we count the number of times it came up
    
    
    stat_p = stats.chisquare(freq).pvalue
    if(stat_p < 0.01 ):
        print(f'on the {j:3d}th trial, p-value={stat_p}')
````

Chances are that if you run the code cell above, you will find a few p-values below the threshold of 0.01, which would lead us to reject the null-hypothesis that the dice is fair. This is an example of a type I error discussed at the start of this demo since we are pretty sure the randint function will be provide a random output.

Let's recall what a p-value is: the probability of observing values as extreme or more than the sample, assuming the null hypothesis is true. By selecting a threshold of 0.01, for the p-value and trying this experiment many times, we are making a statement about the expected rate of false positives. In fact, given we have 600 experiments in total, with a threshold value of 0.01, we expect to have around 6 false positives. There are two takeaways here:
1. Defining a threshold for significance comes at a cost, there will be a non-null chance of having a type I error (false positive).
2. You should not "test-shop", test everything until something significant comes out. A threshold of 0.05 means that for every twenty tests you run, you will have one false positive.

Sometimes, you have to run several tests in a row. This is ok if you define a threshold *a priori* from the test you perform and you do not change the threshold after seeing the results and if you apply [corrections](https://en.wikipedia.org/wiki/Bonferroni_correction) to the threshold. 

This demonstration can also be fine-tuned to showcase a type II error: a false negative. Instead of a dice, we will do it a biased coin: it lands on "heads" 26/50 times and on "tails" 24/50 times. The hypothesis are:
- H$_0$: the coin is fair
- H$_1$: the coin is unfair

We will repeat the experiment several times however, we will only count the number of times we find a p-value **below** 0.01, meaning we are counting the number of times we would reject the null-hypothesis:

````python
faces = 2
tries = 100
trials = faces*tries
n_samples = 2000

count = 0
for j in range(0,trials):
    a = random.choices([0,1],weights=[54/100,46/100],k=n_samples) # we define an unbalanced random output: the unfair dice

    freq = list()
    freq.append(n_samples - np.sum(a)) # the number of times 0 appears = n_samples minus times one appears
    freq.append(np.sum(a)) # the number of times 1 appears is equal to summing all the 1 values in a

    stat_p = stats.chisquare(freq).pvalue
    if(stat_p < 0.01 ):
        count += 1      # we reject the null-hypothesis in this case

print(count)
print(f'we fail to reject the null-hypothesis {count} in {trials} times') 
````

Doing several trials, we still do not reject the null hypothesis every time! This is a type II error, a false negative. This test cannot distinguish an unbalanced coin with probabilities (54/100,46/100) everytime. 

Why does this happen? In short, bad luck. The longer answer is that by setting the threshold p-value at 0.01, we are making another statement about what will it take to convince us to reject the null hypothesis. The bias of the coin is small 54/100 vs 46/100 and with the threshold also being "small", we are saying we will need a lot of evidence before rejecting the null hypothesis. 

To drive this home, let's do this same test again for several threshold levels:

````python
faces = 2
tries = 100
trials = faces*tries
n_samples = 2000

for p_tresh in [1E-2,1E-3,1E-4,1E-5,1E-6]:
    count = 0
    for j in range(0,trials):
        a = random.choices([0,1],weights=[54/100,46/100],k=n_samples) # we define an unbalanced random output: the unfair dice
    
        freq = list()
        freq.append(n_samples - np.sum(a)) # the number of times 0 appears = n_samples minus times one appears
        freq.append(np.sum(a)) # the number of times 1 appears is equal to summing all the 1 values in a
    
        stat_p = stats.chisquare(freq).pvalue
        if(stat_p < p_tresh ):
            count += 1
    print(f'For a threshold of {p_tresh:2.1e}, we reject the null-hypothesis {count:3d} out of {trials:3d} times')
````

The number of times the null-hypothesis is (correctly) rejected goes down! 

This means picking a threshold for the p-value is a statement with two sides:
1. The lower the threshold, the stricter we need the evidence to be in order to reject the null hypothesis (Type II error)
2. The higher the threshold, the easier we might convince ourselves to reject the null hypothesis too many times (Type I error)

Note that type I errors are a rate, independent of the number of samples: for a threshold of 0.01, we get approximately 6 false positives. While type II errors are introduced because our threshold for rejection is too strict in relation to the size of the effect we want to observe, considering the number of samples. 

Lastly, as a consequence of that last sentence, we cannot detect type II errors after data collection if we do not know the ground truth about the distribution we are observing. In other words, to have a chance of detecting a type II error, we need to know **what is the size of the effect we are observing** and collect enough samples to observe it. This is a key point in particle physics where thousands or millions of events must be observed to detect one or two which are of interest.

So, how could we make a more systematic analysis of p-values? We will not, but we will complement it. To reject or fail to reject the null-hypothesis is always a statement about values: equality of means, variances, distributions or relationships between variables. In this demo, we have checked that the average quality of white and red wines is different. We will make that claim more explicitly now with **bootstrapping**.

The idea of **bootstrapping** is that we resample our data (with replacement), recompute the statistic of interest and use the spread of values to estimate the uncertainty. We can use this technique to calculate a confidence interval for a statistic:

````python
res_white = stats.bootstrap((white_qual,),np.mean,confidence_level=0.95)
res_red = stats.bootstrap((red_qual,),np.mean,confidence_level=0.95)

print(np.mean(white_qual),", 95% CI [",res_white.confidence_interval.low,res_white.confidence_interval.high,"]")
print(np.mean(red_qual),", 95% CI [",res_red.confidence_interval.low,res_red.confidence_interval.high,"]")
print(stats.ttest_ind(white_qual,red_qual,equal_var=False).pvalue)
````

The code cell above bootstraps the distribution of white and red wine qualities to calculate a confidence interval for the mean with a level of 95%. The 95% here does not mean the probability that the true value is inside this interval. Rather, it means that if you repeat this experiment 100 times, 95 of those the value will be inside this interval. Reporting these numbers together: means, their confidence intervals and the p-value of the t-test provides a much stronger claim than reporting them separately. You can use the same bootstrapping method to estimate uncertainties of parameters. 

````python
def linear_regression(x,y):
    results = stats.linregress(x, y)
    return results.slope, results.intercept

x = df['flipper_length_mm']
y = df['body_mass_g']

a, b = linear_regression(x,y)

res = stats.bootstrap((x,y),linear_regression,confidence_level=0.95,paired=True)
conf_int = res.confidence_interval
low = conf_int.low
high = conf_int.high

print("slope:     ",a,"[",low[0],",",high[0],"]")
print("intercept: ",b,"[",low[1],",",high[1],"]")
````

Which is also useful when using statistical modelling to analyze effects of variables on one another. If you are wondering whether an effect is real but very small or it is just very small and it sampling further could produce a different signal, you could use bootstrapping to see if the confidence interval crosses zero. If it doesn't, the effect you see might be small, but real.

# Modelling pipeline

The whole course we tried to emphasize the creation of scientific narratives through the use of statistical models. In this segment we are explicitly synthesizing every step involved in this process.

1. Research question in terms of an hypothesis
2. Definition of methodology: choice of fitting protocol, model and cross-validation strategy
3. Modelling results, uncertainty on parameters and p-values
4. Limitations of the model
5. Conclusions: Interpretation and/or prediction

You should include at least some of each of these steps in your final project. 

### Research question in terms of an hypothesis
This is probably hard to do at first and it does not come naturally if you are unsure about what could be the final result of your project. It also might happen that your first hypothesis is not a good one. All of this is fine. It is normal that you will cycle through one or two hypothesis as research questions before finding the one that you will have answered.

### Choice of fitting protocol, model and cross-validation strategy
Some of these choices are easy, some of them are more complicated. In my opinion, the most important part in this step is having a good cross-validation strategy. If you do, everything else will fall easier into place.

### Modelling results, uncertainty on parameters and p-values
You should present your result as fully as possible in the most convincing way.

### Limitations of the model
We all know the model, data, etc... have limitations. It is important to go beyond just recognizing these exist. You should be able to name a few limitations, explain why they are recognized as such. Going even further is being able to recognize in precise terms what would need to change such that you can answer your research question definitely.

### Conclusions: Interpretation and/or prediction
Models can serve many purposes, you should keep in mind what is a potential use for the model as a function of your research question. This should inform how you discuss your results before they are shared to the wider world. This is a question of recognizing who is your audience when communicating results. Other scientists? A supervisor? Having this knowledge and understanding will help you highlight the results that matter the most.
